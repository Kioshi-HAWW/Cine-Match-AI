# MovieLens EDA

This notebook performs exploratory data analysis on the MovieLens dataset using Matplotlib. It includes dataset counts, genre and rating distributions, top-rated movies, missing value analysis, and correlation analysis.

```python
import os
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

sns.set_style("whitegrid")

DATA_DIR = Path("..") / "archive" / "ml-latest-small"
FIGURES_DIR = Path("..") / "notebooks" / "figures"
FIGURES_DIR.mkdir(parents=True, exist_ok=True)

movies = pd.read_csv(DATA_DIR / "movies.csv")
ratings = pd.read_csv(DATA_DIR / "ratings.csv")
tags = pd.read_csv(DATA_DIR / "tags.csv")
links = pd.read_csv(DATA_DIR / "links.csv")

print("Movies head:")
display(movies.head())
print(movies.dtypes)

print("Ratings head:")
display(ratings.head())
print(ratings.dtypes)

print("Tags head:")
display(tags.head())
print(tags.dtypes)

print("Links head:")
display(links.head())
print(links.dtypes)
```
